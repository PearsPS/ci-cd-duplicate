```python
import pandas as pd

df = pd.read_csv("data/raw_dataset.csv")

print("Original Dataset")
display(df)

cleaned_df = df.drop_duplicates()

print("Processed Dataset")
display(cleaned_df)

cleaned_df.to_csv("data/processed_dataset.csv", index=False)

print("Duplicate removal completed.")
```
